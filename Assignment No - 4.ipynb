```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("C:/Users/dhira/Downloads/Bappa.jpg", cv2.IMREAD_GRAYSCALE)

plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis("off")
plt.show()

plt.hist(img.ravel(), bins=256, range=(0, 256), color='black')
plt.title("Histogram - Original")
plt.show()
```

```python
# Gaussian Noise
mean = 0
sigma = 20
gaussian = np.random.normal(mean, sigma, img.shape).astype(np.float32)
gaussian_noisy = img.astype(np.float32) + gaussian
gaussian_noisy = np.clip(gaussian_noisy, 0, 255).astype(np.uint8)

plt.imshow(gaussian_noisy, cmap='gray')
plt.title("Gaussian Noise")
plt.axis("off")
plt.show()

plt.hist(gaussian_noisy.ravel(), bins=256, range=(0, 256), color='black')
plt.title("Histogram - Gaussian Noise")
plt.show()
```

```python
# Salt & Pepper Noise
prob = 0.02
salt_pepper_noisy = img.copy()
rnd = np.random.rand(*img.shape)
salt_pepper_noisy[rnd < prob/2] = 0
salt_pepper_noisy[rnd > 1 - prob/2] = 255

plt.imshow(salt_pepper_noisy, cmap='gray')
plt.title("Salt & Pepper Noise")
plt.axis("off")
plt.show()

plt.hist(salt_pepper_noisy.ravel(), bins=256, range=(0, 256), color='black')
plt.title("Histogram - Salt & Pepper Noise")
plt.show()
```

```python
# Impulse Noise
prob = 0.02
impulse_noisy = img.copy()
h, w = img.shape
num_noisy = int(prob * h * w)
coords = [np.random.randint(0, i - 1, num_noisy) for i in img.shape]
for i in range(num_noisy):
    impulse_noisy[coords[0][i], coords[1][i]] = np.random.randint(0, 256)

plt.imshow(impulse_noisy, cmap='gray')
plt.title("Impulse Noise")
plt.axis("off")
plt.show()

plt.hist(impulse_noisy.ravel(), bins=256, range=(0, 256), color='black')
plt.title("Histogram - Impulse Noise")
plt.show()
```
